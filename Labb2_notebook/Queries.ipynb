```sql
SELECT
Students.FirstName + ' ' + Students.LastName AS 'Name',
Staff.FirstName + ' ' + Staff.LastName AS 'Mentor'
FROM Students
JOIN Classes ON Students.ClassId = Classes.ClassId
JOIN Staff ON Classes.MentorId = Staff.StaffId;
```
