```sql
CREATE PROCEDURE ShowStaffWithPosition 
    @PositionParam INT
AS
BEGIN
    SELECT * 
    FROM Staff 
    WHERE Position = @PositionParam;
END;
```

```sql
use School;
SELECT 
Subjects.SubjectName,
AVG(Grades.Credit) AS 'Avg Credit' ,
MIN(Student_Grades.GradeLevel) AS 'Highest Grade',
MAX(Student_Grades.GradeLevel) AS 'Lowest Grade'
FROM Subjects
JOIN Student_Grades ON Student_Grades.SubjectId = Subjects.SubjectId
JOIN Grades ON Student_Grades.GradeLevel = Grades.Grade
GROUP BY SubjectName;
```

```sql
CREATE PROCEDURE AddStudent 
    @FirstName VARCHAR(50),
    @LastName VARCHAR(50),
    @PersonalNumber VARCHAR(50),
    @ClassId INT
AS
BEGIN
INSERT INTO Students(FirstName, LastName, PersonalNumber, ClassId)
VALUES
(@FirstName, @LastName, @PersonalNumber, @ClassId)
END;
```

```sql
-- stored procedure för att lägga till nya elever
EXEC AddStudent 'Hugo', 'Wakt', '050302-9023', 1;
```
